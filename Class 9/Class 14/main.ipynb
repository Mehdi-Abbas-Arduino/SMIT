```python
from datetime import datetime
class Person:
    def __init__(self,name,gender,dob) -> None:
        self.name = name
        self.gender = gender
        self.dob = dob 

    def find_age(self):
        birthyear = int(self.dob.split('-')[0])
        current_year = datetime.now().year
        age = current_year - birthyear
        print(age)

per1 = Person("Mehdi","M","2008-03-01")
```

```python
per1.find_age()
```

```python
class Child(Person):
    def __init__(self, name, gender, dob,id) -> None:
        super().__init__(name, gender, dob)
        self.id = id 
    def course(self,coursename):
        print(f"The student {self.name} has {coursename}")

child1 = Child("Mehdi",'M','2008-03-01',256789)
```

```python
child2 = Child("Salman Mehdi Abbas","M","2030-03-01",256789)
```

```python
child1.find_age()
```

```python
child1.course("Python")
```

```python
class Animal:
    def __init__(self, name) -> None:
        self.name = name

    def speak(self):
        print("Animal speaks")

class Bird:
    def __init__(self, color) -> None:
        self.color = color

    def fly(self):
        print("Bird flies")

class Bat(Animal, Bird):
    def __init__(self, name, color) -> None:
        Animal.__init__(self, name)
        Bird.__init__(self, color)

    def sleep(self):
        print("Bat sleeps during the day")

bat = Bat("Tom", "Black")
bat.speak()  
bat.fly()    
bat.sleep()  
```

```python
Bat.name = "Sam"
```

```python
Bat.name
```

```python
class FederalGovernment:
    def __init__(self,country):
        self.country = country
    def  get_country(self):
        print(f"Federal Government of Pakistan")
```

```python
class ProvincialGovernment(FederalGovernment):
    def __init__(self, name, budget):
        super().__init__(name)
        self.budget = budget    
```

```python
class LocalGovernment(ProvincialGovernment):
    def __init__(self, name,budget, population, area):
        super().__init__(name,budget)
        self.population = population
        self.area = area
```

```python
a = LocalGovernment("KMC","5000000","2000000000","Karachi")
```

```python
a.budget
```

```python
class doctor:
    level = "Senior"
    def __init__(self, name, age, specialty):
        self.name = name
        self.age = age
        self.specialty = specialty
```

```python
class assistant(doctor):
    def __init__(self, name, age, specialty):  
        super().__init__(name,age,specialty)
    def  prescribe_medication(self, patient, medication):
        print(f"Dr. {self.name} has prescribed {medication} to {patient}")
```

```python
class supervisior(doctor):
    def __init__(self, name, age, specialty):  
        super().__init__(name,age,specialty)
    def  supervise(self, patient):
        print(f"{self.name} is supervising {patient.name}")
```

```python
Hina = doctor("Hina",67,"MBBS")
```
