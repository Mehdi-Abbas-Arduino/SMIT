```python
lis = [2, 3, 4, 5, 6, 7]  
for num in lis:
    if num == 2:
        print(f"{num} is Prime")
    else:
        is_prime = True
        for j in range(2, num):  
            if num % j == 0:     
                print(f"{num} is Not Prime")
                is_prime = False
                break            
        if is_prime:
            print(f"{num} is Prime")
```

```python
class BankAccount:
    title = "Bank Al Habib"
    No_of_customers = 0
    def __init__(self, account_number, account_name, balance=0):
        self.account_number = account_number
        self.account_name = account_name
        self.balance = balance
        BankAccount.No_of_customers += 1
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount  # Update the balance
            print(f"The new balance is: {self.balance}")
            return self.balance
        else:
            return "Deposit amount must be positive"

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds"
        elif amount <= 0:
            return "Withdrawal amount must be positive"
        else:
            self.balance -= amount  # Update the balance
            print(f"The new balance is: {self.balance}")
            return self.balance
    def transfer(self, amount, recipient_account):
        if amount <= 0:
            return "Transfer amount must be positive"
        elif amount > self.balance:
            return "Insufficient funds for transfer"
        else:
            self.withdraw(amount)
            recipient_account.deposit(amount)
            print(f"Transferred {amount} from {self.account_name} to {recipient_account.account_name}")
```

```python
obj1 = BankAccount(123456,"Mehdi Abbas",200000)
```

```python
obj1.deposit(5000)
```

```python
obj1.withdraw(5000)
```

```python
obj2 = BankAccount(12456789,"Naqi",200000)
```

```python
obj1.transfer(2000,obj2)
```
