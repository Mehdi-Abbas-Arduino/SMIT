# Encapsulation
1. Private : Nor it can be use in derived class and in it's own classes
2. Protected :  Accessible within the class and its derived classes
3. Public :  Public members are accessible from any where in the program.

```python
class BankAccount:
    def __init__(self, account_number, account_name, balance=0):
        self.account_number = account_number
        self.account_name = account_name
        self.balance = balance
    def get_balance(self):
        print(f"Your balance is {self.balance}")
    def deposit(self):
        amount = float(input("Enter the amount to deposit: "))
        self.balance += amount
        print(f"Deposited {amount}. Your new balance is {self.balance}")
    def withdrawal(self):
        amount = float(input("Enter the amount to withdraw: "))
        self.balance = self.balance - amount
        print(f"Withdrew {amount}. Your new balance is {self.balance}")
```

```python
Al_Habib = BankAccount(123456789,"Mehdi",5000)
```

```python
Al_Habib.get_balance()
```

```python
Al_Habib.deposit()
```

```python
Al_Habib.withdrawal()
```

```python
class base:
    def  __init__(self, a,b,y):
        self.a = a
        self._b = b # singlee underscore protected
        self.__y = y # double  underscore private
    def gety(self):
        return self.__y
class derived(base):
    def __init__(self, a, b,y):
        base.__init__(self,a, b,y)
    def get_data(self):
        print(self.a,self._b)
object1 = derived(1,2,3)
```

```python
object1.get_data()
```

# Abstraction

```python
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def payment_process(self, amount, payer_name):
        if amount <= 0:
            print("Invalid amount. Payment process aborted.")
            return False
        if not payer_name:
            print("Payer's name is required. Payment process aborted.")
            return False
        return True

class Debit(Payment):
    def payment_process(self, amount, payer_name):
        print(f"Debit payment process started for {payer_name}. Amount: ${amount}")

class OnlinePayment(Payment):
    def payment_process(self, amount, payer_name):
        print(f"Online payment process started for {payer_name}. Amount: ${amount}")

class Crypto(Payment):
    def payment_process(self, amount, payer_name):
        print(f"Crypto payment process started for {payer_name}. Amount: ${amount}")
```

# Polymorphism

```python
class employee:
    def __init__(self, name, age, salary):
        self.name = name
        self.age = age
        self.salary = salary
```

```python
class full_time(employee):
    def __init__(self, name, age, salary):
        super().__init__(name,age,salary)
    def tax (self):
        return self.salary * 0.10
```

```python
class remote(employee):
    def __init__(self, name, age, salary):
        super().__init__(name, age, salary)
    def tax(self):
        return (self.salary * 0.05)
```

```python
Mehdi = remote("Mehdi",16,2000)
```

```python
Mehdi.tax()
```
